The main parameter of choice for the concept explanations are ``n_concepts``, the number of concepts (clusters) that should be discovered, ``snippet_len`` the number of tokens that should be used as the length of the instances in one cluster. A lower length makes the clusters more coherent, but is also less nuances (mostly words that reappear). Longer lengths can be seemingly less meaningful, but can have a higher level of abstraction...
